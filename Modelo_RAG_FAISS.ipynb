# **Retrieval Augmented Generation (RAG)**

**Retrieval Augmented Generation (RAG)** é uma técnica que combina a capacidade de geração de texto de Large Language Models (LLMs) com a recuperação de informações de fontes externas. O objetivo principal do RAG é mitigar as limitações dos LLMs, como a falta de conhecimento sobre eventos recentes ou domínios específicos não presentes em seus dados de treinamento ou a tendência a "alucinar".


## Funcionamento

Um sistema RAG opera em duas fases principais:

1.  **Fase de Recuperação (Retrieval):** Quando uma pergunta ou prompt é fornecido, o sistema primeiro recupera documentos ou trechos de texto relevantes de uma base de conhecimento externa (por exemplo, um banco de dados vetorial, uma coleção de documentos, a web). Esta base de conhecimento é geralmente indexada e pesquisável.
2.  **Fase de Geração (Generation):** Os documentos recuperados são então passados para o LLM como **contexto adicional**, juntamente com a pergunta original. O LLM utiliza este contexto para gerar uma resposta mais informada, precisa e factual.

## Componentes Chave  

*   **Indexador (Indexer):** Processa a base de conhecimento (documentos) e a armazena em um formato otimizado para busca, geralmente embeddings vetoriais em um **banco de dados vetorial**.
*   **Embedder (Gerador de Embeddings):** Um modelo (geralmente um modelo de linguagem menor e mais eficiente) que converte texto (documentos e consultas) em vetores numéricos de alta dimensão (embeddings). Eles capturam o **significado semântico do texto**.
*   **Recuperador (Retriever):** Recebe a consulta do usuário, a converte em um embedding e a compara com os embeddings dos documentos indexados para encontrar os mais semanticamente similares.
*   **Gerador (Generator):** O LLM que recebe a consulta do usuário e os documentos recuperados e gera a resposta final.


# Vantagens do RAG

*   **Redução de Alucinações:** O LLM é "ancorado" em informações factuais, diminuindo a probabilidade de gerar respostas incorretas.
*   **Conhecimento Atualizado:** A base de conhecimento pode ser atualizada independentemente do LLM, permitindo que o sistema acesse informações recentes sem a necessidade de retreinar o modelo completo.
*   **Transparência e Auditabilidade:** É possível rastrear as fontes de informação usadas pelo LLM para gerar uma resposta, aumentando a confiança e a capacidade de auditoria.
*   **Conhecimento Específico de Domínio:** Permite que LLMs genéricos respondam a perguntas sobre domínios altamente especializados, simplesmente adicionando documentos relevantes à base de conhecimento.
*   **Custo-benefício:** Geralmente mais barato e rápido do que o fine-tuning para incorporar novo conhecimento.

# RAG vs. Fine-tuning

Ambas as técnicas – RAG e Fine-tuning – são usadas para adaptar LLMs a tarefas específicas ou a novos dados, mas elas servem a propósitos ligeiramente diferentes e são complementares.

| Característica         | RAG (Retrieval Augmented Generation)                               | Fine-tuning                                                               |
| :--------------------- | :----------------------------------------------------------------- | :------------------------------------------------------------------------ |
| **Objetivo Principal** | Fornecer ao LLM acesso a **novas informações ou informações atualizadas** para gerar respostas factuais e específicas do domínio. | Adaptar o **estilo, tom, formato de saída e capacidades de raciocínio** do LLM para uma tarefa específica. |
| **Mecanismo**          | Recupera documentos relevantes e os injeta como **contexto de entrada** para o LLM.                                                              | Ajusta os **pesos internos** do LLM usando um dataset de exemplos de tarefa.                                                                     |
| **Conhecimento**       | Acesso a **conhecimento externo**, dinamicamente atualizável.       | Incorpora **novo conhecimento nos pesos do modelo** (mais difícil de atualizar).                                                               |
| **Custo/Complexidade** | Geralmente mais simples e menos custoso computacionalmente. Requer uma boa base de conhecimento e um sistema de recuperação. | Mais custoso computacionalmente (especialmente para modelos grandes). Requer dataset de fine-tuning de alta qualidade e infraestrutura de GPU. |
| **Flexibilidade**      | Alta flexibilidade para mudar o domínio do conhecimento sem retreinar o LLM.                                                                | Menor flexibilidade para mudar o domínio sem novo fine-tuning.             |
| **Alucinações**        | Reduz significativamente, pois o LLM tem fontes para consultar.     | Pode reduzir se o fine-tuning for em dados de alta qualidade, mas o modelo ainda pode alucinar.                                               |
| **Casos de Uso**       | Chatbots de suporte ao cliente, sistemas de perguntas e respostas sobre documentos específicos, pesquisa de informações, geração de conteúdo factual. | Análise de sentimento personalizada, sumarização de estilo específico, tradução, geração de código, adaptação a um tom de voz de marca. |

> **Importante notar que RAG e fine-tuning não são mutuamente exclusivos e eles podem ser usados em conjunto!**

## 1. Base de Dados Própria

```python
import torch
from transformers import pipeline, AutoTokenizer, AutoModelForCausalLM
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np

documents = [
    "Brazil is the largest country in South America, known for its vast Amazon rainforest and stunning beaches. Its capital is Brasília.",
    "The Amazon rainforest is the world's largest tropical biome, critical for global biodiversity and climate. It spans nine countries.",
    "Brasília was inaugurated on April 21, 1960, and serves as the federal capital of Brazil. It is famous for its modernist architecture by Oscar Niemeyer.",
    "Oscar Niemeyer was a Brazilian architect, one of the great names in modern architecture. He designed many buildings in Brasília and worldwide.",
    "Artificial Intelligence (AI) is a field of computer science dedicated to developing systems capable of performing tasks that normally require human intelligence.",
    "Machine Learning is a subfield of AI that allows systems to learn from data without being explicitly programmed. It is the foundation of many recent AI advances.",
    "Deep Learning is a category of Machine Learning that uses artificial neural networks with multiple (deep) layers to learn data representations. It is fundamental for LLMs.",
    "Large Language Models (LLMs) are Deep Learning models trained on vast amounts of text to understand, generate, and respond to human language."
]
```

## 2. Fase de Embedding e Indexação

```python
print("--- 1. Criando Embeddings e Índice FAISS ---")

# Modelo Leve para Embeddings (Melhor para CPU e RAG)
embedding_model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')

# 1. Gerar Embeddings para os Documentos
document_embeddings = embedding_model.encode(documents, convert_to_numpy=True)
dimension = document_embeddings.shape[1] # Dimensão do vetor (384 para MiniLM)

# 2. Construir o Índice FAISS
# IndexFlatL2 é o índice mais simples, baseado na distância euclidiana (L2)
index = faiss.IndexFlatL2(dimension)
index.add(document_embeddings.astype('float32'))

print(f"Índice FAISS criado com {index.ntotal} documentos, dimensão: {dimension}")
```

## 3. Fase de Geração Aumentada (RAG)

É natural pensar que os modelos de embedding e de geração precisam ser os mesmos, mas na arquitetura RAG, eles são intencionalmente modelos diferentes pois têm com propósitos distintos! Os modelos têm responsabilidades completamente separadas:

1. A tarefa do primeiro é traduzir textos em vetores numéricos de alta qualidade (representação semântica);

2. A tarefa do segundo é receber o texto do prompt aumentado (que contém o contexto recuperado e a pergunta) e gerar uma resposta coerente e em linguagem natural.

> **O segundo modelo NUNCA LÊ O VETOR!!!**. O embedding gerado nunca alimenta o segundo modelo que só tem acesso aos documentos recuperados!


```python
from transformers import pipeline, AutoTokenizer, AutoModelForQuestionAnswering

# --- Configuração do Modelo QA ---
LLM_MODEL_NAME_QA = "distilbert-base-uncased-distilled-squad"
LLM_TASK = "question-answering"

# Carregar o pipeline de Question Answering
qa_pipeline = pipeline(
    LLM_TASK,
    model=LLM_MODEL_NAME_QA,
    tokenizer=LLM_MODEL_NAME_QA
)

def run_rag_query_qa(query: str, k: int = 2):
    # --- 1. Retrieval (FAISS e Embeddings) ---

    # 1. Geração do Embedding da Query
    query_embedding = embedding_model.encode([query], convert_to_numpy=True)
    query_embedding = query_embedding.astype('float32')

    # 2. Busca no Índice FAISS
    distances, indices = index.search(query_embedding, k)

    # 3. Recuperação dos Documentos
    retrieved_docs = [documents[i] for i in indices[0]]
    context = " ".join(retrieved_docs) # Combine os documentos em um único contexto

    print(f"\n--- Query: '{query}' ---")
    print(f"Contexto Usado: {context[:100]}...") # Exibe um trecho do contexto

    # --- 2. Geração Aumentada (Usando o Pipeline de QA) ---

    # O pipeline de QA usa a função 'context' e 'question'
    output = qa_pipeline(
        question=query,
        context=context
    )

    # O output do QA é um dicionário contendo a resposta, score e posição
    generated_text = output['answer']

    return generated_text

```

## 4. Execução das Perguntas

```python
# Pergunta 1
response_1 = run_rag_query_qa("Who designed the capital of Brazil and when was it inaugurated?", k=2)
print("\n✅ Resposta RAG 1:")
print(response_1)

# Pergunta 2
response_2 = run_rag_query_qa("LLMs uses machine learning?", k=2)
print("\n✅ Resposta RAG 2:")
print(response_2)
```

Se estivéssemos usando um LLM Generativo (como GPT-2, Llama ou Gemini) conseguiríamos responder aqui às perguntas sem contexto (zero-shot). O distilbert-base-uncased-distilled-squad (QA), entretanto é modelo QA extrativo, e não consegue gerar texto livremente.

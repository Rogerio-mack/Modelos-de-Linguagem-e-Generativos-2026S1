# SQuAD XLM-RoBERTa

```python
from transformers import pipeline

# O XLM-RoBERTa fine-tuned para SQuAD-v2 suporta várias línguas, incluindo português.
print("Carregando o modelo de Question-Answering...")
qa_pipeline = pipeline("question-answering", model="deepset/xlm-roberta-base-squad2")
print("Modelo carregado com sucesso!")

# Definir a pergunta e um contexto relevante para a resposta
question = "Qual a cor do céu?"
context = "Durante o dia, o céu geralmente aparece azul devido à dispersão da luz solar. Ao nascer e pôr do sol, pode assumir tons de laranja, rosa e roxo. Em dias nublados, o céu pode parecer cinza."

# Executar o modelo para obter a resposta
print(f"\nPergunta: {question}")
print(f"Contexto: {context}")

result = qa_pipeline(question=question, context=context)

# Imprimir a resposta do modelo
print(f"\nResposta do modelo: {result['answer']}")
print(f"Score de confiança: {result['score']:.2f}")
```

# SQuAD bert-base-cased-squad-v1.1-portuguese

```python
# Instalar dependências
!pip install transformers torch --quiet

# Importar bibliotecas
from transformers import pipeline

# Criar pipeline de Question Answering
qa_pipeline = pipeline(
    "question-answering",
    model="pierreguillou/bert-base-cased-squad-v1.1-portuguese",
    tokenizer="pierreguillou/bert-base-cased-squad-v1.1-portuguese"
)

# Contexto (importante para QA!)
contexto = """
O céu é geralmente azul durante o dia devido à dispersão da luz solar na atmosfera.
Ao pôr do sol, ele pode apresentar tons de vermelho, laranja e rosa.
"""

# Pergunta
pergunta = "Qual a cor do céu?"

# Executar QA
resposta = qa_pipeline(
    question=pergunta,
    context=contexto
)

# Exibir resultado
print("Pergunta:", pergunta)
print("Resposta:", resposta["answer"])
print("Score de confiança:", resposta["score"])
```
